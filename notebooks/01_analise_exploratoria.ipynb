```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
customers = pd.read_csv(ROOT / 'data/processed/customer_segments.csv')
profiles = pd.read_csv(ROOT / 'data/processed/segment_profiles.csv')
model_selection = pd.read_csv(ROOT / 'data/processed/model_selection.csv')
customers.shape, profiles.shape
```

## Qualidade e completude

```python
customers[['monthly_income', 'asset_total', 'credit_exposure', 'product_count']].describe().T
customers.isna().mean().sort_values(ascending=False).head(10)
```

## Comparação de K

```python
display(model_selection.round(4))
sns.lineplot(data=model_selection, x='k', y='silhouette', marker='o')
plt.title('Silhouette por número de clusters')
plt.show()
```

K=5 preserva cinco personas acionáveis, embora soluções menores tenham maior silhouette. A decisão combina métricas e utilidade de negócio.

```python
cols = ['segment', 'customers', 'customer_share', 'avg_age', 'avg_income', 'avg_assets', 'avg_credit_utilization', 'avg_digital_share']
display(profiles[cols].sort_values('customers', ascending=False).round(3))
```

```python
sample = customers.sample(2500, random_state=42)
plt.figure(figsize=(10, 6))
sns.scatterplot(data=sample, x='monthly_income', y='investment_balance', hue='segment', alpha=.6)
plt.title('Renda e investimentos por segmento')
plt.show()
```

## Uso responsável

Os segmentos apoiam hipóteses e personalização. Não devem restringir crédito, determinar capacidade financeira ou substituir revisão humana e políticas validadas.
